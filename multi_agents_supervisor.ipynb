```python
from langchain_community.tools import DuckDuckGoSearchRun
def search_duckduckgo(query :str):
  """searched internet using duckduckgo"""
  search = DuckDuckGoSearchRun()
  return search.invoke(query)

result = search_duckduckgo("what are ai agents")
print (result)
```

```python
def add(a:float,b:float)->float:
  """add two number"""
  return a+b

def multiply(a:float,b:float)->float:
  """multiply two numbers"""
  return a*b
```

```python
from langgraph_supervisor import create_supervisor
from langgraph.prebuilt import create_react_agent

math_agent = create_react_agent(
    model = model,
    tools=[add,multiply],
    name = "math expert",
    prompt="you are a math expert,always use one tool at a time."
)
```

```python
research_agent = create_react_agent(
    model = model,
    tools=[search_duckduckgo],
    name = "research expert",
    prompt="you are a research expert,always use one tool at a time,do not do any math"
)
```

```python
workflow = create_supervisor(
    [research_agent,math_agent],
    model = model ,
    prompt=(
        "you are a team suppervisor managing a research expert and a math expert"
        "for current events use research_agent"
        "for math problem use math agent"
    )
)
```

```python
app = workflow.compile()

result = app.invoke({
    "messages":[
        {
            "role":"user",
            "content":"what is quantum computing?, try answering using tools only"
        }
    ]
})
```
